```python
import os

def check_unique_file_names(dir_path, end_extension=".csv"):
    """
    Check if there are any duplicate file names in a directory.

    Args:
        dir_path (str): The path to the directory to check.
        end_extension (str): The file extension to check for.

    Returns:
        bool: True if there are no duplicate file names, False otherwise.
    """

    files = [f for f in os.listdir(dir_path) if f.endswith(end_extension)]
    unique_files = set(files)
    
    return len(unique_files) 
```

```python
import os

def check_min_series_len(data_folder):
    """
    Check the shortest time series length in a folder of CSV files.
    Returns:
        min_length (int): The length of the shortest series.
        min_file (str): The name of the file with the shortest series.
    """
    min_length = float('inf')
    min_file = None

    for file in os.listdir(data_folder):
        if file.endswith(".csv"):
            file_path = os.path.join(data_folder, file)
            with open(file_path, 'r') as f:
                lines = f.readlines()
                length = len(lines)
                if length < min_length:
                    min_length = length
                    min_file = file

    return (min_length, min_file) if min_file else (0, None)

check_min_series_len("/home/umair/Betterdata Environment/longtimeseries/fractalgen/raw_data/train")
```

```python
import os

def remove_short_series(data_folder, min_required_length=2000):
    """
    Removes .csv files in the folder with fewer than `min_required_length` lines.
    
    Args:
        data_folder (str): Path to the folder with CSV files.
        min_required_length (int): Minimum required length of a series.
        
    Returns:
        deleted_files (list): List of file names that were deleted.
    """
    deleted_files = []

    for file in os.listdir(data_folder):
        if file.endswith(".csv"):
            file_path = os.path.join(data_folder, file)
            with open(file_path, 'r') as f:
                lines = f.readlines()
                if len(lines) < min_required_length:
                    os.remove(file_path)
                    deleted_files.append(file)

    return deleted_files
```
